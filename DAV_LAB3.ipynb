```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("libraries imported successfully")
```

```python
data={"Student":["Asha","Bharat","Chitra","Deepak","Esha","Farhan",
                 "Gowri","Harish","Isha","Jatin","Kavya","Lokesh",
                 "Meera","Nikhil","Pooja","Rahul","Sneha","Tarun","Uma","Varun"],
      "Department":["CSE","AIML","CSE","ISE","AIML","CSE","ISE"
      ,"CSE","AIML","ISE","CSE","AIML",
                    "ISE","CSE","AIML","ISE","CSE","AIML","ISE","CSE"],
      "Attendance":[92,78,85,96,88,72,91,65,87,82,94,76,89,97,84,69,93,80,86,90],
      "Study_Hours":[3.5,2,3,5,4,1.5,4.5,1,3.5,2.5,5.5,2,4,6,3,1.5,5,2.5,3.5,4.5],
      "Assignment_Score":[82,68,78,94,86,62,90,55,84,73,96,70,88,98,79,60,93,74,85,91],
      "Internal_Marks":[84,70,80,95,87,65,91,58,85,75,97,72,89,99,81,63,94,76,86,92],
      "Final_Score":[86,72,82,96,88,66,93,60,86,77,98,73,90,99,83,64,95,78,87,94]}
df=pd.DataFrame(data)
display(df.head())
print("Shape:",df.shape)
print(df.dtypes)
```

HISTOGRAM

```python
plt.figure(figsize=(8,5))
plt.hist(df["Final_Score"],bins=7,edgecolor="green")
plt.title("Distribution of final scores")
plt.xlabel("Final_Score");plt.ylabel("Number of students")
plt.show()
```

BOX PLOT

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Department",y="Final_Score")
plt.title("Distribution of final scores by department")
plt.xlabel("Deapartment Name");plt.ylabel("Final_Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Study_Hours"],df["Final_Score"])
plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours");plt.ylabel("Final Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=80)
plt.title("Study Hours vs Final Score by Departmant")
plt.show()
```

PAIR PLOT

```python
num_clos=["Attendance","Study_Hours","Assignment_Score","Internal_Marks","Final_Score"]
sns.pairplot(df[num_clos])
plt.show()
```

```python
corr=df[num_clos].corr()
display(corr)
```

```python
plt.figure(figsize=(9,6))
sns.heatmap(corr,annot=False,fmt=".2f",cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
movie_data={"Movie":["Sky Force","Code Red","Midnight Run","Ocean Echo","City Lights","Last Game","Silent Road","Future World","Hidden Truth","Wild Journey","Digital Dreams","Final Mission"],"Budget_Crore":[80,120,60,45,35,70,50,150,40,90,110,130],"Duration_Minutes":[145,160,125,135,110,150,130,170,120,155,165,158],"Rating":[8.2,7.5,8,7.8,6.9,8.5,7.6,7.2,8.1,7.9,7.4,8.3],"Revenue_Crore":[240,280,150,110,75,260,130,310,140,220,290,350]}
movies=pd.DataFrame(movie_data)
movies
```

HISTOGRAM

```python
movies[["Rating", "Budget_Crore"]].hist(figsize=(10,4))
plt.show()

```

SCATTER PLOT

```python
sns.scatterplot(data=movies, x="Budget_Crore", y="Revenue_Crore")
plt.show()
```

```python
sns.pairplot(movies[["Budget_Crore", "Duration_Minutes", "Rating", "Revenue_Crore"]])
plt.show()
```

```python
sns.heatmap(movies.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

THREE DATA-DRIVEN OBSERVATIONS

```python
corr = movies.corr(numeric_only=True)

print("1. Budget and Revenue have a strong positive correlation of",
      round(corr.loc["Budget_Crore", "Revenue_Crore"], 2))

print("2. Rating and Revenue have a moderate positive correlation of",
      round(corr.loc["Rating", "Revenue_Crore"], 2))

print("3. Budget and Duration have a strong positive correlation of",
      round(corr.loc["Budget_Crore", "Duration_Minutes"], 2))

```
